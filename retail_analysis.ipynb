```python
import pandas as pd

# 1. Load the raw dataset from your folder path
file_path = r"C:\Users\Nathan\Documents\Projects Data\online_retail_II.xlsx"

print("Loading raw Excel sheets...")
df_2009 = pd.read_excel(file_path, sheet_name="Year 2009-2010")
df_2010 = pd.read_excel(file_path, sheet_name="Year 2010-2011")

# Combine both years into a single dataframe
df_raw = pd.concat([df_2009, df_2010], ignore_index=True)
print(f"Loaded raw rows: {len(df_raw):,}")

# 2. Clean and structure the data
df_raw.rename(columns={'Customer ID': 'CustomerID'}, inplace=True)
df_raw['Description'] = df_raw['Description'].fillna('UNKNOWN')
df_raw['InvoiceDate'] = pd.to_datetime(df_raw['InvoiceDate'])

# Remove invoice cancellations (starting with 'C') and negative quantities
df_clean = df_raw[~df_raw['Invoice'].astype(str).str.startswith('C') & (df_raw['Quantity'] > 0)].copy()

# Save the fresh clean master file
df_clean.to_csv('retail_clean_master.csv', index=False)
print(f"SUCCESS! Cleaned dataset saved. Total clean rows: {len(df_clean):,}")
```
