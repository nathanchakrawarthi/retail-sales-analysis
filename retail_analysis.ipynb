```python
from openpyxl import load_workbook

wb = load_workbook(file_path, read_only=True, data_only=True)

for ws in wb.worksheets:
    cancelled = 0
    negative_qty = 0

    for row in ws.iter_rows(
        min_row=2, min_col=1, max_col=4, values_only=True
    ):
        invoice = row[0]
        quantity = row[3]

        if isinstance(invoice, str) and invoice.startswith("C"):
            cancelled += 1

        if isinstance(quantity, (int, float)) and quantity < 0:
            negative_qty += 1

    print(ws.title)
    print("Cancellation invoice rows:", cancelled)
    print("Negative quantity rows:", negative_qty)

wb.close()
```

```python
from openpyxl import load_workbook

wb = load_workbook(file_path, read_only=True, data_only=True)

for ws in wb.worksheets:
    sales = 0
    returns = 0

    for row in ws.iter_rows(
        min_row=2, min_col=4, max_col=6, values_only=True
    ):
        quantity, price = row[0], row[2]

        if quantity is None or price is None:
            continue

        amount = quantity * price

        if quantity > 0:
            sales += amount
        elif quantity < 0:
            returns += amount

    print(f"\n{ws.title}")
    print("Positive-quantity revenue:", round(sales, 2))
    print("Negative-quantity adjustments:", round(returns, 2))
    print("Net revenue:", round(sales + returns, 2))

wb.close()
```

```python
from openpyxl import load_workbook

wb = load_workbook(file_path, read_only=True, data_only=True)
country_revenue = {}

for ws in wb.worksheets:
    for row in ws.iter_rows(min_row=2, values_only=True):
        quantity = row[3]
        price = row[5]
        country = row[7]

        if quantity is None or price is None or country is None:
            continue

        amount = quantity * price
        country_revenue[country] = country_revenue.get(country, 0) + amount

wb.close()

top_10 = sorted(
    country_revenue.items(),
    key=lambda item: item[1],
    reverse=True
)[:10]

for country, revenue in top_10:
    print(country, "—", round(revenue, 2))
```

```python
import csv
from openpyxl import load_workbook

file_path = r"C:\Users\Nathan\Documents\Projects Data\online_retail_II.xlsx"
output_path = r"C:\Users\Nathan\Documents\Projects Data\retail_sql_sample.csv"

wb = load_workbook(file_path, read_only=True, data_only=True)

headers = [
    "Invoice", "StockCode", "Description", "Quantity",
    "InvoiceDate", "Price", "CustomerID", "Country",
    "Period"
]

with open(output_path, "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.writer(f)
    writer.writerow(headers)

    for ws in wb.worksheets:
        period = ws.title

        for row_num, row in enumerate(
            ws.iter_rows(min_row=2, values_only=True), start=2
        ):
            if (row_num - 2) % 100 != 0:
                continue

            row = list(row)

            if row[4] is not None:
                row[4] = row[4].strftime("%Y-%m-%d %H:%M:%S")

            writer.writerow(row + [period])

wb.close()

print("SQL sample created successfully!")
print(output_path)
```
